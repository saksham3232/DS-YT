# Numpy Indexing and Slicing of Vectors

```python
arr = np.arange(11, 21)
arr
```

```python
arr[9]
```

```python
arr[3::2]
```

# NUMPY Indexing and Slicing of Matrix

```python
arr = np.arange(1, 31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[5]
```

```python
arr[0,0]
```

```python
arr[0,4]
```

```python
arr
```

```python
slice = arr[1:3, 1:3]
slice
```

```python
arr[3:, 3:]
```

```python
arr[:, 2]
```

# Boolean Indexing

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

# Heat Map

```python
flights = sns.load_dataset('flights')
flights
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tip_corr = tips[['total_bill', 'tip', 'size']]
```

```python
tip_corr
```

```python
tip_corr.corr()
```

```python
sns.heatmap(tip_corr.corr(), annot=True)
```

# Cluster Map

```python
sns.clustermap(tip_corr.corr(), annot=True)
```

# pivot table heat map

```python
pivotflight = flights.pivot_table(index='month',  columns='year', values='passengers')
pivotflight
```

```python
sns.heatmap(pivotflight, cmap='magma', linecolor='white', linewidth=1)
```
